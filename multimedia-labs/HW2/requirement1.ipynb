```python
import numpy as np
import matplotlib.colors as mcolors
from matplotlib import cm
import matplotlib.pyplot as plt
```

```python
matrix = []
matrix = np.arange(0, 1 + (1 / 255), 1 / 255)
```

```python
def pic(color_map, norm = None, extend = None):
    if norm is None:
        norm = mcolors.Normalize(vmin = 0, vmax = color_map.N)
        
    image = cm.ScalarMappable(norm = norm, cmap = color_map)
    figure,ax = plt.subplots(figsize = (7, 1))
    figure.subplots_adjust(bottom = 0.1)
    figure.colorbar(image, cax = ax, orientation = 'horizontal', extend = extend)
    plt.show()
```

```python
colors = []
for i in range(len(matrix)):
    r = 0 + matrix[i]
    g = 1 - r
    b = 1
    colors.append([r, g, b])

color_map = mcolors.ListedColormap(colors)
print("Requirement1 result：")
pic(color_map)
```
