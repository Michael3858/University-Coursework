```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from PIL import Image
```

```python
image = cv2.imread('./hw2_2.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imwrite('./Gray.jpg', image)
```

```python
image_gray = cv2.imread('./Gray.jpg', 0)
clahe = cv2.createCLAHE(clipLimit = 2.0, tileGridSize = (1000, 1000))
cl = clahe.apply(image_gray)
cv2.imwrite('./bin.jpg', cl)
```

```python
img_list = cv2.imread('./bin.jpg')
b, g, r = cv2.split(img_list)
for i in range(len(img_list)):
    r[i] = 0 + r[i]
    g[i] = 255 - r[i]
    b[i] = 255
img_rgb = cv2.merge([b, g, r])
cv2.imwrite('./Virtual.jpg', img_rgb)
```

```python
image1 = cv2.imread('./Gray.jpg', 0)
image2 = cv2.imread('./Virtual.jpg', 0)
plt.subplot(121)
plt.hist(image1.ravel(), 256)
plt.subplot(122)
plt.hist(image2.ravel(), 256)
plt.show()
```
